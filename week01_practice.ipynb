```python
import csv
from pathlib import Path

# Create our practical workspace directory
workspace_dir = Path("/content/week01")
workspace_dir.mkdir(parents=True, exist_ok=True)

raw_file_path = workspace_dir / "raw_customers.csv"

# Generate a raw CSV file with messy data
raw_data = [
  ["customer_id", "name", "email"],
  ["1", " john doe ", "JOHN@example.com"],
  ["2", "jane smith", "jane.smith@Example.com"],
  ["3", " bob ", "bob@example.com "]
]

with open(raw_file_path, mode="w", newline="") as file:
  writer = csv.writer(file)
  writer.writerows(raw_data)

print(f"Created test file at: {raw_file_path}\n")

# Read the CSV file into Python
customers = []
with open(raw_file_path, mode="r") as file:
  reader = csv.DictReader(file) # DictReader uses the first row as dictionary keys
  for row in reader:
    customers.append(row)

print("Raw Data Read from CSV:")
for customer in customers:
  print(customer)
```

```python
cleaned_customers = []

# Clean the records
for c in customers:
  clean_name = c["name"].strip().title()
  clean_email = c["email"].strip().lower()

  cleaned_customers.append({
      "customer_id": c["customer_id"],
  "name": clean_name,
  "email": clean_email
  })

print("Cleaned Data Preview:")

for c in cleaned_customers:
  print(c)

# Write cleaned records to a new CSV file
clean_file_path = workspace_dir / "cleaned_customers.csv"

with open(clean_file_path, mode="w", newline="") as file:
  fieldnames = ["customer_id", "name", "email"]
  writer = csv.DictWriter(file, fieldnames=fieldnames)

  writer.writeheader()
  writer.writerows(cleaned_customers)

print(f"\nSuccess! Cleaned data saved to {clean_file_path}")
```

```python
def validate_email(email):
    if "@" in email:
        return True
    else:
        return False

# Test the function with the requested examples
test1 = "invalid-email.com"
test2 = "valid@example.com"

print(f"Is '{test1}' valid? {validate_email(test1)}")
print(f"Is '{test2}' valid? {validate_email(test2)}")
```
